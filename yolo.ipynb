```python
import torch
print("Torch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())
print("Device count:", torch.cuda.device_count())
```

```python
from pathlib import Path
import zipfile

# Folder where your notebook is
base = Path(".").resolve()

# Zip file in the same directory as the notebook
zip_path = base / "Skin-Problem-Detection-Multiple-Clean.v1i.yolov8.zip"

# Where to extract
root = base / "skin_dataset"

if not root.exists():
    with zipfile.ZipFile(zip_path, "r") as z:
        z.extractall(root)
        print("Extracted dataset to:", root.resolve())
else:
    print("Using existing dataset at:", root.resolve())

data_yaml = root / "data.yaml"
print("data.yaml exists:", data_yaml.exists(), "| path:", data_yaml)
```

```python
from ultralytics import YOLO
model = YOLO('yolov8s.pt')
```

```python
from pathlib import Path
root = Path('./face_skin_yolo').resolve() # root dataset path

results = model.train(
    data=str(data_yaml),
    epochs=30,
    imgsz=512,
    batch=8,
    device=0,
    project="runs",
    name="skin_yolo_run",
    seed=0,
    patience=20,
    verbose=True
)
```

```python
from pathlib import Path
import shutil

best_weights = Path("runs/detect/skin_yolo_run/weights/best.pt")
target_weights = Path("skin_yolo.pt")

if best_weights.exists():
    shutil.copy(best_weights, target_weights)
    print("Saved best model as:", target_weights.resolve())
else:
    raise FileNotFoundError(f"Could not find {best_weights}. Did training finish successfully?")
```
